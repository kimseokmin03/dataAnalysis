```python
import os
from pathlib import Path
from dotenv import load_dotenv

load_dotenv(Path('..') / '.env')
FRED_API_KEY = os.getenv('FRED_API_KEY')
if not FRED_API_KEY:
    raise RuntimeError('FRED_API_KEY is not set. Copy .env.example to .env')
import yfinance as yf
import pandas as pd
from fredapi import Fred

fred = Fred(api_key=FRED_API_KEY)
start_date = '2000-01-01'
end_date = '2025-11-30'
periods = 1 
```

```python
ticker = ['QQQ', 'XLP', 'XLY', 'GC=F']
ydata = yf.download(ticker, start=start_date, end=end_date)['Close'].ffill(limit=7)
#ydata.rename(columns={'IYR':'home'},inplace=True)
ydata
```

```python
ticker = 'CSUSHPINSA'
home = fred.get_series(ticker, start_date, end_date).resample('D').ffill(limit=31)
home.name = 'home'
home = pd.DataFrame(home)
home
```

```python
finance_raw = pd.concat([ydata, home],axis=1).ffill(limit=7)
finance_raw.dropna(inplace=True)   
#finance_raw = finance_raw.pct_change(periods=periods,fill_method=None)*100
finance_raw
```

```python
finance_raw.index.name = 'Date'
finance_raw = finance_raw.reset_index()
#finance_raw.drop(columns='Ticker', inplace=True)
finance_raw.dropna(inplace=True)
finance_raw.head()
```

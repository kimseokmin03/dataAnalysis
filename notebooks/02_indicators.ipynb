```python
import os
from pathlib import Path
from dotenv import load_dotenv

load_dotenv(Path('..') / '.env')
FRED_API_KEY = os.getenv('FRED_API_KEY')
if not FRED_API_KEY:
    raise RuntimeError('FRED_API_KEY is not set. Copy .env.example to .env')
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
from fredapi import Fred
import yfinance as yf

fred = Fred(api_key=FRED_API_KEY)
periods = 365
```

```python
start_date='2014-08-01'
end_date= '2025-12-01'
```

```python
ticker = 'FEDFUNDS'
rate = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#rate = rate.diff(periods=365)
rate
```

```python
ticker = 'DGS10'
government = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#government = government.diff(periods=365)
government = government.resample("MS").first()
government
```

```python
ticker = 'DX=F'
dolar2 = yf.download('DX=F', start=start_date, end = end_date)['Close']['DX=F']
dolar2 = dolar2.resample('MS').first()
dolar2
```

```python
ticker = 'CPIAUCSL'
cpi = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#cpi = cpi.pct_change(periods=365) * 100
cpi
```

```python
ticker = 'DCOILWTICO'
oil = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#oil = oil.pct_change(periods=365) * 100
oil=oil.resample("MS").first()
oil
```

```python
ticker = 'GDPC1'
gdp = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#gdp = gdp.pct_change(periods=365) * 100
gdp = gdp.pct_change().resample("MS").ffill().bfill()
gdp
```

```python
ticker = 'PAYEMS'
hire = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#hire = hire.pct_change(periods=365) * 100
hire
```

```python
ticker = 'VIXCLS'
fear = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#fear = fear.pct_change(periods=365) * 100
fear=fear.resample("MS").first()
fear
```

```python
ticker = 'UNRATE'
fire = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#fire = fire.pct_change(periods=365) * 100
fire
```

```python
ticker = 'M2SL'
msupply = fred.get_series(ticker, start_date, end_date)#.resample('D').ffill()
#fire = fire.pct_change(periods=365) * 100
msupply
```

```python
ticker = 'PCEPILFE'
pce = fred.get_series(ticker,start_date,end_date)
pce
```

```python
final_df = pd.DataFrame({
    'rate': rate,#금리
    'government' : government,#미국 10년국채 수익률
    'dolar' : dolar2,#달러 인덱스
    'cpi': cpi,#cpi
    'gdp': gdp,#gdp
    'oil': oil,#유가
    'hire' : hire,#고용률
    'fear' : fear,#시장공포지수
    'fire' : fire,#실업률
    'msupply' : msupply
})
#final_df = final_df.ffill(limit=10)

diff_columns = ['rate','government','fear', 'fire']
pct_columns = ['cpi','gdp','hire','oil','msupply','dolar'] # +'dolar'

for columns in diff_columns :
        final_df[columns] = final_df[columns].diff()

for columns in pct_columns:
        final_df[columns] = final_df[columns].pct_change()

final_df.index = pd.to_datetime(final_df.index)
#final_df = final_df.set_index('Date')
final_df
```

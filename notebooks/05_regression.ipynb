```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RidgeCV
```

```python
X_df = pd.read_csv('../data/indicator.csv', index_col=0, parse_dates=True)
Y_raw_df = pd.read_csv('../data/finance_raw.csv', index_col=0, parse_dates=True)
```

```python
Y_df = Y_raw_df['QQQ']
final_data = X_df.join(Y_df, how='inner').dropna().sort_index()

x = final_data.drop('QQQ', axis=1)
y = final_data['QQQ']


split_idx = int(len(final_data) * 0.8)
X_train, X_test = x.iloc[:split_idx], x.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

alphas_to_try = np.logspace(-6, 6, 100)
ridge_cv = RidgeCV(alphas=alphas_to_try)

ridge_cv.fit(X_train_scaled, y_train)
print(f"선택된 알파: {ridge_cv.alpha_:.4f}\n")

test_score = ridge_cv.score(X_test_scaled, y_test)
print(f"R^2: {test_score:.4f}\n")

coefficients = pd.Series(ridge_cv.coef_, index=x.columns)
print(coefficients.sort_values(ascending=False))
```
